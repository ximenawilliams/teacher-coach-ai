```python
import pandas as pd

# cargar curriculum configuration
curriculum = pd.read_csv(
    "/kaggle/input/datasets/ximenawilliams08/ds-coach/admin_curriculum_example.csv"
)

# cargar student performance data
students = pd.read_csv(
    "/kaggle/input/datasets/ximenawilliams08/ds-coach/student_scores.csv"
)


print("Curriculum Dataset")
print(curriculum.head())

print("\nStudent Scores Dataset")
print(students.head())

# Detect students with learning gaps
risk_students = students[students["Score"] < 70]

risk_students

## Contexto curricular

curriculum_context = ""

for _, row in curriculum.iterrows():
    curriculum_context += f"""
Objetivo: {row['Objetivos']}
Contenidos: {row['Contenidos']}
Indicadores: {row['Indicadores']}
Actividades: {row['Actividades']}
---
"""

print(curriculum_context)

## estudiantes

student_case = risk_students.iloc[0]

prompt = f"""
You are Teacher Coach AI.

Use the following curriculum context to generate an educational recommendation.

Curriculum context:
{curriculum_context}

Student performance:
Student: {student_case['Student']}
Grade level: {student_case['Grade']}
Subject: {student_case['Subject']}
Topic: {student_case['Topic']}
Competency: {student_case['Competency']}
Score: {student_case['Score']}
Attendance: {student_case['Attendance']}
Objective: {student_case['Objective']}
Resources: {student_case['Resources']}
Teaching method: {student_case['Teaching_Method']}
Evaluation type: {student_case['Evaluation_Type']}

Generate:
1. Learning gap
2. Reinforcement recommendation
3. Suggested classroom activity
4. Teacher guidance

Keep it practical and short.
"""

print(prompt)


from kaggle_secrets import UserSecretsClient
import google.generativeai as genai

# Load secret manager
user_secrets = UserSecretsClient()

# Get API key from Kaggle Secrets
GOOGLE_API_KEY = user_secrets.get_secret(GOOGLE_API_KEY)

# Configure Gemma/Gemini
genai.configure(api_key=GOOGLE_API_KEY)

# Load model
model = genai.GenerativeModel(
    "gemma-3-1b-it"
)

# Generate recommendation
response = model.generate_content(prompt)

print(response.text)
```
